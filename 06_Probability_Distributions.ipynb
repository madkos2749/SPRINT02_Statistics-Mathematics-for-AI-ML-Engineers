PROBABILITY DISTRIBUTIONS

DEFINITION:
A probability distribution shows all the possible values an event can take and the probability (chance) of each value occurring.

EXAMPLE:
 
For a dice:

Outcome:       1   2   3   4   5   6
Probability:  1/6 1/6 1/6 1/6 1/6 1/6
The probability distribution tells us that each number has a 1/6 chance of appearing.

BERNOULLI DISTRIBUTION

DEFINITION:

Bernoulli Distribution is a probability distribution used when there are only two possible outcomes for an experiment.

Bernoulli distribution answers a simple Yes/No, True/False, or Success/Failure question.

FORMULA:
              Let:

X = 1 → Success

X = 0 → Failure

p → probability of success

1-p → probability of failure

The probability mass function is:

                                  P(X=x)=px(1−p)1−x

        where:

                      x∈{0,1}

          So:

P(X=1)=p

P(X=0)=1−p

Bernoulli Distribution Used in ?

Useful whenever there are two possible outcomes, such as:

1. Email → Spam / Not Spam

2. Customer → Buy / Not Buy

3. Loan → Default / No Default

4. Machine → Failure / No Failure

5. Patient → Disease / No Disease

6. Advertisement → Click / No Click

Real-World Example:

STUDENT ATTENDANCE
Suppose a Student attends class or not.

The student either:

Attends Class     → (1) Success
Doesn't attend     → (0) Failure

```python
import numpy as np

purchase = np.random.binomial(n=1, p=0.7)

print("Purchase:", purchase)
```

We can visualize a Bernoulli distribution using a simple bar chart.

```python
import matplotlib.pyplot as plt

# Probability of success
p = 0.2

outcomes = [0, 1]
probabilities = [1 - p, p]

plt.bar(outcomes, probabilities, color=["skyblue", "orange"])

plt.xticks([0, 1], ["Failure", "Success"])
plt.ylabel("Probability")
plt.title("Bernoulli Distribution")

plt.show()
```

NOTE:
Bernoulli Distribution = Yes/No, 0/1, Success/Failure.

BINOMIAL DISTRIBUTION

DEFINITION:

Binomial Distribution is used when we perform the same experiment multiple times, where each trial has only two possible outcomes, such as Success/Failure or Yes/No.

In simple words:

Binomial Distribution tells us the probability of getting a certain number of successes in a fixed number of independent trials.

FORMULA:

                      P(X=k)=(nk)pk(1−p)n−k

 Where:

n = total number of trials

k = number of successes

p = probability of success in one trial

1-p = probability of failure

X = number of successes

Binomial Distribution Used in ?

It is useful when we have:

A fixed number of trials

Only two possible outcomes per trial

The same probability of success

Independent trials

Common applications include:

Advertisement clicks

Customer purchases

Product defects

Loan approvals

Medical test results

Email spam detection

Website conversions

EXAMPLE:

```python
import numpy as np

present = np.random.binomial(n=15, p=0.2)
print("Student who attends Classes are :", present)
```

NOTE:

Binomial Distribution = probability of getting a certain number of successes in a fixed number of independent Yes/No trials.

POISSON DISTRIBUTION:

DEFINITION:
Poisson Distribution is used to find the probability of getting a certain number of events within a fixed time, distance, area, or space, when those events happen independently at an average rate.

In simple words:

Poisson Distribution tells us how many times something is likely to happen in a fixed period or area.

EXAMPLES:
1. Number of customers arriving in 1 hour

2. Number of website requests in 1 minute

3. Number of machine failures in 1 month

4. Number of accidents at a particular location in a day

FORMULA:

P(X=k)=e^-λλkk!

Where:

X = number of events

k = number of events we are interested in

λ (lambda) = average number of events in the given interval

e ≈ 2.718

k! = factorial of k

For example, if a food-delivery app receives an average of 4 orders per minute, then:

  λ=4

The Poisson Distribution can tell us the probability of receiving exactly 6 orders in one minute.

EXAMPLE:

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import poisson

# Average number of events
lam = 5

# Possible number of events
x = np.arange(0, 16)

# Poisson probabilities
probability = poisson.pmf(x, lam)

plt.bar(x, probability, color="skyblue", edgecolor="black")

plt.xlabel("Number of Customers")
plt.ylabel("Probability")
plt.title("Poisson Distribution: Coffee Shop Customers")
plt.xticks(x)
plt.show()
```

NOTE: 
Poisson Distribution = probability of a certain number of events happening in a fixed interval when we know the average event rate.

UNIFORM DISTRIBUTION:

DEFINITION:

Uniform Distribution is a probability distribution where all possible values have the same chance of occurring within a given range.

In simple words:

Uniform Distribution means every value in a given range is equally likely.

For example, if a fair random number is generated between 1 and 10, each number has the same chance of being selected.

FORMULA:

For a continuous uniform distribution between a and b:

                              f(x)=1/b−a`, a≤x≤b
        Where:

a = minimum value

b = maximum value

x = possible value

The probability density is constant across the range.

For example, between 10 and 20:

f(x)=1/20−10=0.1

So the distribution has a flat shape.



Where is Uniform Distribution Used?

1. Uniform Distribution can be useful for:

2. Random number generation

3. Simulation

4. Random sampling

5. Monte Carlo methods

6. Data augmentation

7. Initializing some ML model parameters

8. Testing algorithms under equally likely conditions

EXAMPLE:

```python
import numpy as np
import matplotlib.pyplot as plt

# Generate 1000 equally distributed random values
data = np.random.uniform(1, 5, 1000)

plt.hist(data, bins=10, edgecolor="black", color="skyblue")

plt.xlabel("Rating")
plt.ylabel("Frequency")
plt.title("Uniform Distribution of Random Ratings")

plt.show()
```

NORMAL DISTRIBUTION:

DEFINITION:

Normal Distribution, also called Gaussian Distribution, is a probability distribution where most data values are close to the average, while fewer values occur as we move farther away from the average.

In simple words:

Normal Distribution is a bell-shaped pattern where most values are near the center and fewer values are at the extremes.

FORMULA:

f(x)=1σ2πe−(x−μ)22σ2

Where:

μ (mu) = mean / average

σ (sigma) = standard deviation

x = data value

π ≈ 3.14159

e ≈ 2.71828

Where is Normal Distribution Used?

Normal Distribution is commonly useful in:

Statistical analysis

Data preprocessing

Feature analysis

Measurement-error modeling

Hypothesis testing

Confidence intervals

Machine learning algorithms that make distributional assumptions

```python
import numpy as np
import matplotlib.pyplot as plt

# Generate normally distributed delivery times
delivery_time = np.random.normal(30, 5, 1000)

plt.hist(delivery_time, bins=30, edgecolor="black",
         color="skyblue", density=True)

plt.xlabel("Delivery Time (minutes)")
plt.ylabel("Density")
plt.title("Normal Distribution of Delivery Times")

plt.show()
```

NOTE:
a bell-shaped distribution where most values are near the average and fewer values occur farther from the average.

EXPONENTIAL DISTRIBUTION:

DEFINITION:

Exponential Distribution is a probability distribution used to model the waiting time until the next event happens, when events occur independently at a constant average rate.

In simple words:

Exponential Distribution tells us how long we may have to wait until the next event occurs.

FROMULA:
                                        f(x)=λe−λx, x≥0
        Where:

λ (lambda) = average event rate

x = waiting time

e ≈ 2.718

Where is Exponential Distribution Used?

It is commonly used for:

Customer waiting times

Call-center waiting times

Website/server request intervals

Machine failure times

Network packet arrival times

Time between medical events

Reliability analysis

EXAMPLE:

```python
import numpy as np
import matplotlib.pyplot as plt

rate = 2

x = np.linspace(0, 5, 1000)

y = rate * np.exp(-rate * x)

plt.plot(x, y, color="blue")

plt.xlabel("Waiting Time (minutes)")
plt.ylabel("Probability Density")
plt.title("Exponential Distribution")

plt.grid()
plt.show()
```

NOTE:
Exponential Distribution = the probability of how long we wait until the next event happens.
