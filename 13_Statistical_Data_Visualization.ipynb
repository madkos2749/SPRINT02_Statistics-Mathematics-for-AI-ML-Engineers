# STATISTICAL DATA VISUALIZATION

### DEFINITION:

Statistical data visualization is the process of representing statistical data in graphical or visual form so that patterns, trends, relationships, distributions, and comparisons can be understood easily.

## HISTOGRAM

### PURPOSE:

A histogram is a graphical representation used to show the distribution of numerical data.

It divides data into intervals called bins and displays how many observations fall within each interval.

### When to USE:

* To understand the distribution of numerical data.
* Want to see whether data is approximately normally distributed.
* Want to identify skewness.
* To detect possible outliers.
* To understand the spread and concentration of data.

### ADVANTAGES:

* Easy to understand.
* Shows the shape of data distribution.
* Helps identify skewness.
* Can reveal possible outliers.

### LIMITATIONS:

* Mainly used for numerical data.
* The appearance depends heavily on the number and width of bins.
* Exact individual values are not easily visible.

### EXAMPLE:
a food delivery company records the delivery time for 20 orders.

```python
import matplotlib.pyplot as plt

delivery_time = [
    20, 22, 25, 28, 30,
    31, 32, 35, 35, 36,
    38, 40, 42, 45, 45,
    48, 50, 52, 55, 60
]

plt.hist(delivery_time, bins=5, edgecolor="black")

plt.xlabel("Delivery Time (minutes)")
plt.ylabel("Number of Orders")
plt.title("Distribution of Delivery Times")

plt.show()
```

## BAR CHART

### PURPOSE:

A bar chart is a graphical representation used to compare values across different categories.

Each category is represented by a rectangular bar, and the height or length of the bar represents its value.

### When to USE:

* To compare categories.
* When data is categorical.
* To compare sales across products.
* To compare employees, regions, departments, or stores.
* To show rankings such as highest to lowest.

### ADVANTAGES:

* Very easy to understand.
* Excellent for comparing categories.
* Clearly shows highest and lowest values.
* Can display values using vertical or horizontal bars.

### LIMITATIONS:

* Not ideal for showing continuous numerical distributions; use a histogram instead.
* Too many categories can make the chart difficult to read.
* Small differences may be difficult to notice if the chart is poorly scaled.

### EXAMPLE:



```python
import matplotlib.pyplot as plt
Stationary = ["Pencils", "Pens", "Sharpners", "Books"]
Quantity = [9999, 15555, 8088, 17000]
plt.bar (Stationary, Quantity)
plt.xlabel ("STATIONARY")
plt.ylabel ("REQUIRED QUANTITY")
plt.title ("REQUIREMENT OF STATIONARY")

plt.show()
```

## LINE CHART:

### PURPOSE:

A line chart is a graphical representation used to show changes or trends in data over time or across an ordered sequence.

It connects data points with lines, making it easy to see whether values are increasing, decreasing, or remaining stable.

### When to USE:

* When to show trends over time.
* When Data is ordered chronologically.
* To compare changes across months, quarters, or years.
* For identify increases or decreases.
* To observe patterns such as seasonal changes.

### ADVANTAGES:

* Clearly shows trends.
* Easy to understand.
* Good for time-series data.
* Makes increases and decreases easy to identify.

### LIMITATIONS:

* Not ideal for comparing many unrelated categories.
* Too many lines can make the chart confusing.

### EXAMPLE:

DAYS & Sales:

```python
import matplotlib.pyplot as plt
Days = ["Sunday", "Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday"]
Sales = [5000, 2000, 27000, 14000, 9000, 6700, 7700]
plt.plot(Days, Sales, marker="o")

plt.xlabel("DAYS")
plt.ylabel("SALES (₹)")
plt.title("DAYS SALES TREND")

plt.show()
```

## SCATTER PLOT

### PURPOSE:

A scatter plot is a graphical representation used to show the relationship between two numerical variables.

Each observation is represented as a point on the graph.

### When to USE:

* To study relationship between two numerical variables.
* To identify Positive or Negative Relationships
* To Detect Outliers.

### ADVANTAGES:

* Simple and easy to understand.
* Clearly shows relationships between two variables.
* Helps identify positive and negative trends.
* Can reveal outliers.

### LIMITATIONS:

* Mainly shows the relationship between two variables at a time.
* With a very large dataset, points can overlap (overplotting).
* It does not prove causation.

### EXAMPLE:

Fruits Investement and Fruits Profit

```python
import matplotlib.pyplot as plt
Fruits_Investment = [70000, 80000, 90000, 50000, 60000, 100000]
Fruits_Profit = [85000, 95000, 110000, 65000, 80000, 127000]
plt.scatter (Fruits_Investment, Fruits_Profit)

plt.xlabel("FRUITS INVESTMENT")
plt.ylabel("FRUITS PROFIT")
plt.title("FRUITS BUSINESS")

plt.show()
```

## PIE CHART

### PURPOSE:

A pie chart is a circular chart used to show how a whole is divided into different categories or parts.

Each category is represented by a slice, and the size of the slice represents its proportion of the total.

### When to USE:

* To show parts of a whole.
* Categories add up to a meaningful total.
* To divide into a small number of categories.
* To show percentages or proportions.

### ADVANTAGES:

* Visually attractive.
* Clearly shows the largest and smallest portions.
* Useful for presenting percentages.

### LIMITATIONS:

* Difficult to compare slices that are very similar in size.
* Too many categories make the chart confusing.
* Exact values are harder to compare than with a bar chart.
* Not suitable for showing changes over time.

### EXAMPLE:



```python
import matplotlib.pyplot as plt
Liquids = ["Water", "Juice", "Cooking oil", "Shower gel"]
Consumed = [890, 9900, 5000, 750]
plt.pie (Consumed, labels = Liquids, autopct="%1.1f%%")
plt.title ("LIQUIDS CONSUMPTION PER MONTH")
plt.show()
```

## DENSITY PLOT

### PURPOSE:

A density plot is a graph used to show the distribution of a numerical variable using a smooth curve.

It is similar to a histogram, but instead of bars, it uses a smooth line to represent where values are concentrated.

### When to USE:
* To understand the distribution of numerical data.
* To identify where values are concentrated.
* To compare the distributions of two or more groups.
* To identify skewness.

### ADVANTAGES:

* Produces a smooth and easy-to-read distribution curve.
* Helps identify the shape of the distribution.
* Useful for comparing distributions between groups.

### LIMITATIONS:

* The curve is an estimate, not the exact distribution of the data.
* Results can change depending on the bandwidth/smoothing used.
* Less intuitive for beginners than a histogram.
* Can hide individual observations.

### EXAMPLE:
An online store records how much customers spend.


```python
import seaborn as sns
import matplotlib.pyplot as plt

spending = [200, 250, 300, 320, 350, 400, 420, 450,
            480, 500, 520, 550, 600, 650, 700]

sns.kdeplot(spending, fill=True)

plt.xlabel("Customer Spending (₹)")
plt.ylabel("Density")
plt.title("Customer Spending Distribution")

plt.show()
```

## BOX PLOT

### PURPOSE:

A box plot, also called a box-and-whisker plot, is a statistical graph used to show the distribution, spread, center, and possible outliers of numerical data.

It summarizes data using five important values:

* Minimum
* Q1 (First Quartile)
* Median
* Q3 (Third Quartile)
* Maximum
The box mainly represents the middle 50% of the data, called the Interquartile Range (IQR).

Simple meaning: A box plot helps us quickly understand the center, spread, and outliers in numerical data.

### When to USE:

* To understand the spread of numerical data.
* To identify outliers.
* To compare distributions between different groups.
* To compare the median of multiple groups.
Your data may not follow a normal distribution.

### ADVANTAGES:


* Clearly shows the median.
* Shows the spread of the middle 50% of observations.
* Helps identify potential outliers.

### LIMITATIONS:

* Does not show the exact shape of the distribution as clearly as a histogram or density plot.
* Can hide details such as multiple peaks or clusters.
* May be difficult for beginners to understand initially.

### EXAMPLE:

A company wants to analyze employee Salaries.

```python
import matplotlib.pyplot as plt

salary = [
    25000, 28000, 30000, 32000, 35000,
    38000, 40000, 42000, 45000, 48000,
    50000, 55000, 60000, 120000
]

plt.boxplot(salary)

plt.ylabel("Salary (₹)")
plt.title("Employee Salary Distribution")

plt.show()
```

## VIOLIN PLOT

### PURPOSE:

A violin plot is a statistical visualization used to show the distribution of numerical data, including its shape, spread, center, and density.

It combines features of a box plot and a density plot.

### When to USE:

* To understand the distribution of numerical data.
* To compare distributions between multiple groups.
* To see where values are highly concentrated.
* To identify whether a distribution is symmetric or skewed.

### ADVANTAGES:

* Shows the shape of the distribution.
* Displays areas where data is highly concentrated.
* Useful for comparing multiple groups.
* Provides more information about distribution than a basic box plot.

### LIMITATIONS:

* More difficult to understand than a bar chart or box plot.
* The density is an estimate, so its shape depends on smoothing/bandwidth.
* Can be misleading with very small datasets.

### EXAMPLE:

Employee Salary by Departments.

```python
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    "Department": [
        "Sales", "Sales", "Sales", "Sales", "Sales",
        "IT", "IT", "IT", "IT", "IT"
    ],
    "Salary": [
        30000, 35000, 40000, 45000, 50000,
        45000, 50000, 55000, 60000, 65000
    ]
}

sns.violinplot(x="Department", y="Salary", data=data)

plt.title("Salary Distribution by Department")
plt.xlabel("Department")
plt.ylabel("Salary (₹)")

plt.show()
```

## PAIR PLOT

### PURPOSE:

A pair plot is a visualization used to examine the relationships between multiple numerical variables at the same time.

It creates a grid of plots where:

* Scatter plots show relationships between pairs of variables.
* Distribution plots on the diagonal show the distribution of each variable.

### When to USE:

* Have several numerical features.
* To explore relationships between variables.
* To identify possible correlations.
* To find potential outliers.

### ADVANTAGES:

* Shows many relationships in one visualization.
* Easy way to perform exploratory data analysis (EDA).
* Helps identify positive and negative relationships.

### LIMITATIONS:

* Can become very large and confusing when there are many features.
* Not ideal for datasets with hundreds of variables.

### EXAMPLE:

 we have information about students' study hours, attendance, assignments, and exam scores.

```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Study_Hours": [2, 3, 4, 5, 6, 7],
    "Attendance": [60, 65, 70, 75, 85, 90],
    "Assignments": [50, 55, 60, 70, 80, 90],
    "Score": [45, 50, 58, 65, 75, 85]
}

df = pd.DataFrame(data)

sns.pairplot(df)

plt.show()
```

## HEAT MAP

### PURPOSE:

A heatmap is a data visualization that uses colors to represent the magnitude or intensity of values in a table or matrix.

Darker or lighter colors can represent higher or lower values, depending on the chosen color scale.

### When to USE:

* To visualize a correlation matrix.
* To identify relationships between multiple variables.
* When we have data arranged in rows and columns.

### ADVANTAGES:

* Easy to identify patterns using colors.
* Can display a large amount of information compactly.
* Makes strong and weak relationships easy to spot.
* Useful for correlation analysis.

### LIMITATIONS:

* Too many variables can make the heatmap difficult to read.
* The choice of color scale can affect interpretation.
* Correlation heatmaps generally show association, not causation.
* Correlation does not necessarily capture nonlinear relationships.

### EXAMPLE:

Suppose a company wants to understand relationships among:

Advertising
Sales
Customers
Website Visits

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    "Advertising": [10, 20, 30, 40, 50],
    "Sales": [30, 40, 55, 70, 85],
    "Customers": [100, 150, 200, 250, 300],
    "Website_Visits": [200, 300, 400, 500, 600]
}

df = pd.DataFrame(data)

# Calculate correlation
corr = df.corr()

# Create heatmap
sns.heatmap(corr, annot=True, cmap="coolwarm")

plt.title("Correlation Heatmap")
plt.show()
```
