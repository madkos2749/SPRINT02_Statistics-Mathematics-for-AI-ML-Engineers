# STATISTICAL TESTS

### DEFINITION:

Statistical tests are mathematical methods used to analyze data and determine whether an observed difference, relationship, or effect is statistically significant or could have occurred due to random chance.

## T-TEST:

### DEFINITION:

A t-test is a statistical test used to determine whether the means (averages) of groups are significantly different.

In AI/ML, a t-test can be used to compare things such as:

* Model performance
* Average prediction errors
* Customer behavior
* Experimental results
* Feature effects

#### Assumptions of T-TEST are:

* Numerical data
* Approximately normal data/differences
* Independent Observations
* No extreme outliers
* Equal Variances

#### When to USE:

* You want to compare means.
* Your dependent variable is numerical.
* You have one or two groups, depending on the t-test type.

#### When NOT to USE:

A t-test may not be appropriate when:

* Your target variable is categorical rather than numerical.
* You have three or more independent groups and want to compare all their means → consider ANOVA.
* Your data has severe outliers.

### EXAMPLE:

Imagine an AI recommendation system in an online shopping app. The company wants to know whether a new recommendation algorithm increases the number of products viewed per user.

Data
Suppose 8 users are tested with the old and new recommendation systems:

Old System: [5, 6, 4, 7, 5, 6, 4, 5]
New System: [7, 8, 6, 8, 7, 9, 6, 8]
Since the same users are tested with both systems, we use a paired t-test.

Hypotheses
H₀: The new recommendation system does not change
    the average number of products viewed.

H₁: The new recommendation system changes
    the average number of products viewed.

```python
from scipy.stats import ttest_rel

old = [5, 6, 4, 7, 5, 6, 4, 5]
new = [7, 8, 6, 8, 7, 9, 6, 8]

# Paired T-Test
t_stat, p_value = ttest_rel(new, old)

print("Old average:", sum(old) / len(old))
print("New average:", sum(new) / len(new))
print("P-value:", round(p_value, 4))

if p_value < 0.05:
    print("Reject H0")
    print("The new recommendation system significantly changes user activity.")
else:
    print("Fail to reject H0")
    print("No significant change detected.")
```

## Z- TEST

### DEFINITION:

A Z-test is a statistical test used to determine whether a sample mean or proportion is significantly different from a known/reference value, or whether two population means/proportions are significantly different.

It uses the standard normal (Z) distribution.

### Assumptions:

* Numerical or proportion data, depending on the specific Z-test.
* Independent observations for the usual independent-sample setting.
* The sampling distribution is approximately normal.
* For a mean Z-test, the population standard deviation (σ) is known, or the sample is large enough for a normal approximation to be reasonable.
* For a proportion Z-test, the sample should be sufficiently large for the normal approximation to work.

#### When to USE:

Use a Z-test when:

* You want to compare a sample mean with a known population mean and σ is known.
* You have a large sample where a normal approximation is appropriate.
* You want to compare proportions, such as conversion rates.

#### When NOT to USE:

A Z-test is generally not the best choice when:

* The sample is small and the population standard deviation is unknown → a t-test is usually more appropriate for a mean.
* The data is strongly non-normal with a small sample.
* Observations are dependent when the test assumes independence.
* You have three or more groups and want to compare their means → consider ANOVA or another appropriate method.
* The outcome/test assumptions don't match a normal approximation.

#### EXAMPLE:

```python
import numpy as np

# Response time of 5 chatbot requests (seconds)
sample = np.array([4.2, 4.5, 4.1, 4.3, 4.4])

# Known population values
population_mean = 5
population_std = 1

# Calculate Z statistic
z = (sample.mean() - population_mean) / (
    population_std / np.sqrt(len(sample))
)

print("Sample Mean:", sample.mean())
print("Z Statistic:", z)
```

## CHI-SQUARE TEST

### DEFINITION:

The Chi-Square (χ²) Test is a statistical test used to determine whether there is a significant relationship between categorical variables or whether observed category frequencies differ from expected frequencies.

It checks whether the observed counts are significantly different from the counts we would expect if the variables were unrelated.

#### Assumptions:

* The variables should be categories such as Yes/No, Male/Female, Purchased/Not Purchased.
* One observation should not influence another.
* The test works with counts, not percentages or continuous measurements.
* Expected frequencies should generally be sufficiently large — A common rule of thumb is that expected counts should be at least 5 in most/all cells. If expected counts are too small, alternatives such as Fisher's exact test may be appropriate.

#### When to USE:

Use a Chi-Square test when you want to investigate:

* Whether two categorical variables are associated.
* Whether observed category frequencies match expected frequencies.
* Whether customer behavior is related to an AI/ML treatment.
* Whether a categorical feature is associated with a categorical target.

#### When NOT to USE:

Don't use a standard Chi-Square test when:

* Your data is continuous, such as salary, temperature, or response time.
* You are comparing the means of numerical data → consider a t-test or ANOVA.
* Observations are not independent.
* Expected cell counts are too small for the Chi-Square approximation to be reliable.
* Your data consists of measurements rather than category counts.

#### EXAMPLE:

```python
import numpy as np
from scipy.stats import chi2_contingency

#                 Purchased   Not Purchased
data = np.array([
    [80,            120],       # Old AI
    [110,            90]        # New AI
])

# Perform Chi-Square Test
chi2, p_value, dof, expected = chi2_contingency(data)

print("Chi-Square Statistic:", round(chi2, 3))
print("P-value:", round(p_value, 4))

print("\nExpected Frequencies:")
print(expected)

alpha = 0.05

if p_value < alpha:
    print("\nReject H0")
    print("There is a significant association.")
else:
    print("\nFail to reject H0")
    print("No significant association was detected.")
```

## ANOVA [Analysis Of Variance]:

### DEFINITION:

ANOVA (Analysis of Variance) is a statistical test used to determine whether the means of three or more groups are significantly different.

### ASSUMPTIONS:

* Independence
* Continuous dependent variable
* Normality
* Random Sampling

### When to USE:

* To Compare the means of 3 or more groups.
* When the groups are independent for a one-way ANOVA.
* The assumptions are reasonably satisfied.

### When NOT to USE:

* The dependent variable is categorical rather than numerical.
* The data have severe outliers or strongly violate assumptions.
* You have only two groups and a simple comparison is sufficient; an independent-samples t-test is commonly used

### EXAMPLE:

A coffee shop wants to know whether three different music styles affect the average number of coffees sold per hour.

They test:

**Group A**: No music
**Group B**: Soft music
**Group C**: Upbeat music

```python
from scipy.stats import f_oneway

no_music = [20, 22, 19, 21, 23]
soft_music = [25, 27, 24, 26, 28]
upbeat_music = [30, 32, 29, 31, 33]

f_stat, p_value = f_oneway(
    no_music,
    soft_music,
    upbeat_music
)

print("F-statistic:", f_stat)
print("P-value:", p_value)

if p_value < 0.05:
    print("There is a significant difference in coffee sales.")
else:
    print("There is no significant difference in coffee sales.")
```

## MANN-WHITNEY U-TEST

### DEFINITION:

The Mann–Whitney U test is a non-parametric statistical test used to compare two independent groups.

It is an alternative to the independent-samples t-test when the data do not satisfy the assumptions required for a t-test, particularly when the data are ordinal or not approximately normally distributed.

Instead of directly comparing means, the Mann–Whitney U test ranks all observations together and compares the ranks between the two groups.

### ASSUMPTIONS:

* Two Independent Groups
* Observations are independent within groups
* Groups are independently sampled
* Data are atleat ordinal.

### When to USE:

* dependent variable is ordinal or continuous.
* When having independent groups

### When NOT to USE:

* more than two independent groups
* Data are paired or repeated measurements

### EXAMPLE:
A food-delivery company wants to know whether Delivery Service A and Delivery Service B have different delivery times.

They record the delivery time in minutes for 6 orders from each service.

```python
from scipy.stats import mannwhitneyu

service_A = [32, 35, 30, 40, 33, 38]
service_B = [25, 28, 27, 29, 26, 24]

u_stat, p_value = mannwhitneyu(
    service_A,
    service_B,
    alternative="two-sided"
)

print("U-statistic:", u_stat)
print("P-value:", p_value)

if p_value < 0.05:
    print("Reject H0: Delivery times are significantly different.")
else:
    print("Fail to reject H0: No significant difference.")
```

## SHAPIRO-WILK TEST

### DEFINITION:

The Shapiro–Wilk test is a statistical test used to determine whether a dataset is approximately normally distributed.

Normality is important because many statistical tests, such as the t-test and ANOVA, make an assumption that the data are approximately normal.

### ASSUMPTIONS:

* Observations are independent
* Data are quantitative
* Observations represent the population/sample of interest

### When to USE:

* to check whether numerical data are approximately normal.
* planning to use a statistical method that assumes normality.

### When NOT to USE:
* dataset is very large.
* data are categorical rather than numerical.

### EXAMPLE:

A shop owner wants to check whether daily ice cream sales are approximately normally distributed before using statistical methods that assume normality.The shop records sales for 10 days:

Sales:
48, 52, 51, 49, 55, 53, 50, 47, 54, 51

```python
from scipy.stats import shapiro

ice_cream_sales = [48, 52, 51, 49, 55, 53, 50, 47, 54, 51]

stat, p_value = shapiro(ice_cream_sales)

print("Shapiro-Wilk statistic:", stat)
print("P-value:", p_value)

if p_value > 0.05:
    print("Fail to reject H0: Sales are approximately normally distributed.")
else:
    print("Reject H0: Sales are not normally distributed.")
```
