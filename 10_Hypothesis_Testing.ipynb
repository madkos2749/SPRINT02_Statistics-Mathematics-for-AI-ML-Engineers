# HYPOTHESIS TESTING

### DEFINITION:

Hypothesis testing helps us decide whether the evidence in our data is strong enough to support a claim.

## NULL HYPOTHESIS:

### DEFINITION:

The Null Hypothesis, written as H₀, is a statement that assumes there is no significant effect, difference, or relationship in an experiment.

Null hypothesis (H₀) means: "Nothing significant has changed."

The Null Hypothesis (H₀) is the default assumption that there is no significant difference or effect.

Example:

The company claims:

Average screen time = 7 hours

So,

H₀ : μ = 7 hours

## ALTERNATIVE HYPOTHESIS

### DEFINITION:

The Alternative Hypothesis, represented by H₁ or sometimes Hₐ, is the hypothesis that says there is a significant difference, effect, or relationship in an AI/ML experiment.

#### EXAMPLE:

Suppose we believe the average screen time is not equal to 10 hours.

H₁ : μ ≠ 7 hours

```python
print("Alternative Hypothesis: average screen time is not 7 hours")
```

## P-VALUE

### DEFINITION:

The p-value is a statistical value used in hypothesis testing to measure how compatible the observed data is with the Null Hypothesis (H₀).

Simple definition
A p-value tells us how surprising our observed result would be if the Null Hypothesis were true.
A small p-value means the observed result would be relatively unusual under H₀, providing evidence against H₀.

A large p-value means the observed result is not unusual enough to provide strong evidence against H₀.

#### FORMULA:
We commonly use:

Significance level (α) = 0.05
Then:

P-value	Decision	Meaning
p < 0.05	Reject H₀	Statistically significant evidence
p ≥ 0.05	Fail to reject H₀	Insufficient evidence of significance

#### EXAMPLE:

Suppose:

p-value = 0.02
α = 0.05
Since:

0.02 < 0.05
we reject H₀.

```python
p_value = 0.03

if p_value < 0.05:
    print("Reject Null Hypothesis")
else:
    print("Fail to Reject Null Hypothesis")
```

## CONFIDENCE INTERVAL

### DEFINITION:

A confidence interval tells us the range within which the true population value is plausibly located, based on a sample of data.

#### EXAMPLE:

Model accuracy = 92%
95% Confidence Interval = [90%, 94%]
Instead of reporting only:

"The model has 92% accuracy."
we can report:

"The estimated accuracy is 92%, with a 95% confidence interval from 90% to 94%."
This communicates the uncertainty around our estimate.

##### Real time Examples:
* Banking: bank evaluates an AI fraud detection model on 10,000 transactions.
* e-commerce company tests a new AI recommendation system

### EXAMPLE:

Battery life confidence interval:

9.9 hours to 10.9 hours

## SIGNIFICANCE LEVEL

### DEFINITION:

The significance level, represented by α (alpha), is a threshold used in hypothesis testing to decide when the evidence against the Null Hypothesis (H₀) is strong enough to reject it.

##### REAL TIME EXAMPLE:
Suppose a company compares two fraud-detection models.

Old model → 90% accuracy
New model → 93% accuracy
They formulate:

H₀: There is no significant difference between the models.

H₁: There is a significant difference.

The data produces:

p-value = 0.03
They selected:

α = 0.05
Comparison:

0.03 < 0.05
Therefore:

Reject H₀.
There is statistical evidence of a difference between the models.

### EXAMPLE:

The most commonly used significance level is:

α = 0.05 This means we accept a 5% significance threshold for rejecting H₀.

```python
alpha = 0.05

print("Significance Level:", alpha)
```

## TYPE-1 ERROR

### DEFINITION:

A Type I Error occurs when we reject the Null Hypothesis (H₀) even though H₀ is actually true.


Type I Error = False Positive

#### EXAMPLE:


```python
import numpy as np
from scipy.stats import ttest_1samp

np.random.seed(7)

# Reality: average performance is actually 80%
data = np.random.normal(loc=80, scale=5, size=30)

# H0: True mean = 80
# H1: True mean != 80

t_stat, p_value = ttest_1samp(data, popmean=80)

alpha = 0.05

print("p-value:", round(p_value, 4))

if p_value < alpha:
    print("Reject H0")
    print("If H0 is actually true, this is a Type I Error.")
else:
    print("Fail to reject H0")
```

##### NOTE:
Type I errors can occur when:

Declaring a new ML model significantly better when it isn't
Declaring a feature useful when it has no real effect
Claiming an A/B test improved conversion when it didn't
Claiming a relationship between variables that doesn't truly exist
Selecting a feature based on a false statistical relationship

## ONE TAILED TEST

### DEFINITION:

A one-tailed test checks whether an observed effect is significantly greater than or significantly less than a specified value in one particular direction.

There are two types:

* **Right-tailed test**: Tests whether the value is greater than the reference value.   H₁ : μ > 10
* **Left-tailed test** : Tests whether the value is less than the reference value.      H₁ : μ < 10

```python
from scipy.stats import ttest_1samp

# Accuracy improvement (%) from 8 experiments
improvement = [2.1, 2.8, 1.9, 3.2, 2.5, 2.7, 1.8, 2.9]

# H0: Average improvement <= 0
# H1: Average improvement > 0

t_stat, p_two_tailed = ttest_1samp(improvement, 0)

# Convert the two-sided p-value to a right-tailed p-value
if t_stat > 0:
    p_one_tailed = p_two_tailed / 2
else:
    p_one_tailed = 1 - (p_two_tailed / 2)

alpha = 0.05

print("Average improvement:", round(sum(improvement) / len(improvement), 2), "%")
print("One-tailed p-value:", round(p_one_tailed, 4))

if p_one_tailed < alpha:
    print("Reject H0")
    print("Evidence suggests the new model improves accuracy.")
else:
    print("Fail to reject H0")
    print("Not enough evidence of improvement.")
```

## TWO TAILED TEST

### DEFINITION:

A two-tailed test checks whether an observed value is significantly different from a reference value, whether the difference is positive or negative.

#### EXAMPLE:

Let's test whether adding a new feature causes a significant change in an ML model's F1-score.

We don't assume whether the score will increase or decrease.

```python
from scipy.stats import ttest_rel

# F1-scores from the same validation runs
without_feature = [0.78, 0.81, 0.79, 0.80, 0.77, 0.82]
with_feature    = [0.83, 0.80, 0.84, 0.78, 0.85, 0.81]

# H0: No difference in performance
# H1: Performance is different

t_stat, p_value = ttest_rel(
    with_feature,
    without_feature
)

alpha = 0.05

print("t-statistic:", round(t_stat, 3))
print("p-value:", round(p_value, 4))

if p_value < alpha:
    print("Reject H0")
    print("The feature causes a statistically significant change.")
else:
    print("Fail to reject H0")
    print("No statistically significant change was detected.")
```

##### NOTE:
A two-tailed test in AI/ML is a hypothesis test used to determine whether an observed parameter, model performance, feature effect, or other measurement is significantly different from a reference value in either direction.

One-tailed → "Is it greater/lower?" Two-tailed → "Is it different?"
