# SAMPLING

### DEFINITION:

Sampling is the process of selecting a smaller, representative set of data points from a large dataset.

Sampling helps to;
1. reduce computation time, and training time.
2. reduce memory/storage requirements.

#### EXAMPLE:
Suppose a list has 500 items of groceries, and to check randomly at counter to weather all the items are packed, a sample of 25 groceries were collected from list.

```python
import random
groceries = list(range(1,500))
sample = random.sample(groceries, 111)
print(f" GROCERIES=: {sample [:12]}" )
```

A sample of 111 groceries are selected from list of 500 groceries.

## RANDOM SAMPLING:

### DEFINITION:

Random sampling is a technique in which data points are selected randomly from a larger dataset, so that each data point has an equal or known chance of being selected.

#### Example:
Suppose we have 20 numbers;

```python
import random
numbers = [3, 7, 13, 17, 23, 27, 33, 37, 43, 47, 53, 57, 63, 67, 73, 77, 83, 87, 93, 97]
sample = random.sample(numbers, 7)
print("Random Sample: ", sample)
```

Here, random.sample() randomly selects 7 numbers from the original list.
20 numbers → randomly select 7 numbers  → sample data

## STRATIFIED SAMPLING:

### DEFINITION:
Dividing the data into groups first, then randomly selecting data from every group.

Stratified Sampling helps to;
1. Easy to implement
2. When properly performed, every record has an equal opportunity to be selected

##### Real-time Examples:

1. Banking: Select customers from different income groups.
2. Education: Select students from different departments or years.

#### Example:

An Villa has differnt blocks as Apartments
*  BLOCK-A 
*  BLOCK-B
*  BLOCK-C
*  BLOCK-D

Owner wants to select 3 houses as winners fromm each BLOCK.

```python
Block_1 = list(range(101, 110))
Block_2 = list(range(201, 210))
Block_3 = list(range(301, 310))
Block_4 = list(range(401, 410))

import random

sample = (
    random.sample(Block_1, 3) +
    random.sample(Block_2, 3) +
    random.sample(Block_3, 3) +
    random.sample(Block_4, 3)
)

print(sample[:12])
```

Every BLOCK is represented equally in the sample.

## CLUSTER SAMPLING:

### DEFINITION:

* Cluster sampling is a sampling technique where a large population is divided into groups called clusters, and then some entire clusters are randomly selected for the study.

* Instead of randomly selecting individual data points, we randomly select groups of data points.

##### Real-time Examples:
* Marketing: Select cities and analyze customers from those cities.

* Government surveys: Select villages or districts and collect information from households.

#### EXAMPLE:
Imagine a company has  Employees data from 4 Companies:

Department A → 100 Employees
Department B → 800 Employees
Department C → 1200 Employees
Department D → 900 Employees

Instead of randomly selecting individual Employees from all Departments, the company randomly selects 2 Departments and analyzes all Employees from those Departments.

This is cluster sampling because the DEPARTMENTS are the clusters.

```python
import random

stores = {
    "Department_A": ["Employee1", "Employee2", "Employee3", "Employee4", "Employee5", "Employee6"],
    "Department_B": ["Employee7", "Employee8", "Employee9", "Employee10"],
    "Department_C": ["Employee1", "Employee2", "Employee3", "Employee4", "Employee5", "Employee6", "Employee7"],
    "Department_D": ["Employee11", "Employee12", "Employee13", "Employee14", "Employee15"]
}

# Randomly select 2 clusters (stores)
selected_Department = random.sample(list(stores.keys()), 2)

# Collect all data from selected clusters
sample = []

for store in selected_Department:
    sample.extend(stores[store])

print("Selected Department:", selected_Department)
print("Sample Data:", sample)
```

##### Code Explanation:

Here:

Department A, B, C, D → Clusters

Store A and D → Randomly selected clusters

All Employees from A and D → Sample

Cluster sampling is a sampling technique in which the population is divided into groups called clusters, and some clusters are randomly selected to form the sample.

## SYSTEMATIC SAMPLING:

### DEFINITION:
Systematic sampling is a sampling technique where we select data points at a fixed regular interval from a dataset after choosing a starting point randomly.

##### Real-time Examples:
* Manufacturing: Inspect every 20th product from a production line.

* Education: Select every 10th student from a student list.

* Healthcare: Select every 5th patient record for analysis.

#### Example:
Suppose we have 88 online orders and want to select every 8th order.

```python
orders = [f"Order_{i}" for i in range(1, 88)]

# Select every n3rd order
sample = orders[2::3]

print(f"Systematic Sample: , {sample}")
```

###### Code Explanation:

The starting point can change, but after that, we consistently select every 4th record.

## SAMPLING BIAS:

### DEFINITION:

Sampling bias happens when the way we choose data causes the sample to be unfair or unrepresentative of the real population.

##### Real-time Examples:

* Banking: Include customers from different income groups.

* Marketing: Include customers with different ages, locations, and preferences.

* AI/ML: Build training datasets that represent the real-world population.

## SAMPLING ERROR:

### DEFINITION:

Sampling error is the difference between the result obtained from a sample and the actual result of the entire population, simply because we studied only a subset of the data.

In simple words: Sampling error happens because a sample may not give exactly the same result as the complete population.

Important: Sampling error is usually due to random variation, not because someone selected the data unfairly. This is different from sampling bias.

##### Real-time Examples:
* Education: Estimate average student marks from a sample of students.

* Polling: Estimate public opinion by surveying a sample of voters.

* AI/ML: Evaluate how well a model performs when only a subset of available data is used.

#### EXAMPLE:

Suppose:

Product average Rating = 4.8, 
Sample average Rating = 5
The difference (0.2) is the sampling error.

```python
product_rating = 4.8
sample_rating = 5.0

sampling_error = sample_rating - product_rating

print(f" Sampling_Error is: {sampling_error}" )
```

##### Code Explanation:
The Difference between Original Product Rating and Collected Sample Rating is defeeined as, Sampling Error. i.e; 0.20000000000000018 for the above example.

## SAMPLING SIZE:

### DEFINITION:

Sample size is the number of data points, observations, or individuals selected from a population for a study or machine-learning task.

In simple words: Sample size tells us how many data points we are going to take from the entire dataset.

##### Real-time Examples:

* Banking: Analyze a sample of transactions for fraud detection.

* Manufacturing: Inspect a certain number of products for quality control.

#### EXAMPLE:
Suppose we have 200000 Employee records in a Company, and CEO wants to select a sample of 121 Employees.

```python
import random

employees = list(range(1, 20001))

sample_size = 121

sample = random.sample(employees, sample_size)

print("Sample Size:", len(sample))
print("Selected Employees:", sample)
```

##### Code Explanation:

This can make experimentation faster and reduce computational requirements.

However, more data is not automatically better if the additional data is poor quality, redundant, or unrepresentative.
