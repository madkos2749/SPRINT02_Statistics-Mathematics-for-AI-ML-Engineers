# FEATURE RELATIONSHIPS

### DEFINITION:

Feature relationships describe how two or more variables (features) in a dataset are related to each other.

## CORRELATION MATRIX

### DEFINITION:
A correlation matrix is a table that shows the correlation coefficients between multiple numerical variables in a dataset.

The most common measure is Pearson correlation, whose value ranges from -1 to +1:

+1 → Perfect positive relationship
0 → No linear relationship
-1 → Perfect negative relationship

### ASSUMPTIONS:

* Numerical Variables
* Linear Relationships
* Independent Observations

### When to USE:
* multiple numerical features.
* exploratory data analysis (EDA).
* want a quick visual summary of many pairwise relationships.

### When NOT to USE:

* variables are purely categorical.
* want to establish causation

### EXAMPLE:

an online learning platform wants to understand the relationship between student study habits and exam performance.

```python
import pandas as pd

data = {
    "Study_Hours": [2, 4, 5, 7, 9],
    "Practice_Tests": [1, 2, 3, 5, 7],
    "Attendance": [60, 70, 75, 85, 95],
    "Exam_Score": [50, 60, 68, 80, 92]
}

df = pd.DataFrame(data)

# Create correlation matrix
correlation_matrix = df.corr()

print(correlation_matrix)
```

## MULTICOLLINEARITY

### DEFINITION:

Multicollinearity occurs when two or more independent variables (features) in a regression or machine-learning model are highly correlated with each other.

### ASSUMPTIONS:

* High correlation between independent variables
* High VIF (Variance Inflation Factor)

### When to USE:

* Building a multiple linear regression model.
* Building a logistic regression model.
* You have many numerical predictors.

### When NOT to USE:

* goal is primarily prediction, and the model performs well.
* You are using models that are relatively robust to correlated features, such as many tree-based models.

### EXAMPLE:

Suppose a restaurant wants to predict daily sales using these features:

Number of Customers
Number of Orders
Number of Tables Used

```python
import pandas as pd

data = {
    "Customers": [20, 30, 40, 50, 60],
    "Orders": [18, 28, 38, 48, 58],
    "Tables": [5, 7, 9, 11, 13],
    "Sales": [5000, 7000, 9000, 12000, 14000]
}

df = pd.DataFrame(data)

# Correlation matrix
print(df.corr())
```

## VARIANCE INFLATION FACTOR [VIF]

### DEFINITION:

Variance Inflation Factor (VIF) is a statistical measure used to detect multicollinearity among independent variables in a regression model.

In simple words:

VIF tells us how much the variance of a regression coefficient is inflated because that predictor is related to the other predictors.

### FORMULA:

                   VIF = 1/(1-R^2)
    WHERE:
         R^2
 is obtained by regressing one predictor against all the other predictors.


A commonly used rule of thumb is:

VIF	Interpretation
1	No multicollinearity
1–5	Usually acceptable
5–10	Potential problem
>10	Serious multicollinearity
These are guidelines rather than strict universal cutoffs.

### ASSUMPTIONS:

* Predictors should be appropriately measured
* Multiple Predictors

### When to USE:

* multiple linear regression model.
* Building a logistic regression model.

### When NOT to USE:

* You have only one independent variable.
* You are calculating relationships between variables without using a regression model.
* Your main goal is simply to calculate correlation.

### EXAMPLE:

a real-estate company wants to predict house prices using:

House Area – square feet
Number of Bedrooms
Number of Rooms

```python
import pandas as pd
from statsmodels.stats.outliers_influence import variance_inflation_factor

data = {
    "Area": [800, 1000, 1200, 1500, 1800],
    "Bedrooms": [2, 2, 3, 4, 4],
    "Rooms": [4, 5, 6, 8, 9]
}

df = pd.DataFrame(data)

vif = pd.DataFrame()
vif["Feature"] = df.columns
vif["VIF"] = [
    variance_inflation_factor(df.values, i)
    for i in range(df.shape[1])
]
print(vif)
```

## FEATURE IMPORTANCE

### DEFINITION:

Feature importance is a technique used in machine learning to determine which input features contribute most to a model's predictions.

### ASSUMPTIOINS:

* The model should be appropriately trained
* Features should be prepared correctly
* Correlated features need attention
* Importance depends on the model.

### When to USE:

* want to know which features are most useful for prediction.
* Want to perform feature selection

### When NOT to USE:

* You need to prove causation.
* Your model performs poorly.

### EXAMPLE:

a company wants to predict whether an employee is likely to leave the company.

The model uses these features:

Job Satisfaction
Monthly Salary
Overtime Hours
Years at Company
Age

```python
import pandas as pd
from sklearn.ensemble import RandomForestClassifier

data = {
    "Satisfaction": [2, 3, 4, 5, 2, 1, 4, 5],
    "Salary": [30, 35, 45, 60, 32, 28, 50, 65],
    "Overtime": [10, 8, 4, 2, 12, 15, 5, 1],
    "Years": [1, 2, 5, 8, 2, 1, 6, 10],
    "Age": [25, 28, 35, 42, 30, 24, 38, 45],
    "Left": [1, 1, 0, 0, 1, 1, 0, 0]
}

df = pd.DataFrame(data)

X = df.drop("Left", axis=1)
y = df["Left"]

model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

model.fit(X, y)

importance = pd.DataFrame({
    "Feature": X.columns,
    "Importance": model.feature_importances_
})

print(importance.sort_values(
    "Importance",
    ascending=False
))
```

## FEATURE SELECTION BASICS

### DEFINITION:

Feature selection is the process of choosing the most useful and relevant features (input variables) from a dataset while removing features that are unnecessary, redundant, or irrelevant.

### ASSUMPTIONS:

* Features should be relevant to the target
* Avoid Data leakage
* Consider Feature redundancy.

### When to USE:

* Dataset has many features.
* Some features are irrelevant.
* Several features contain similar information.
* Your model is too complex.

### When NOT to USE:

* Have very few features and all are potentially useful.
* Remove features without checking their predictive value.

### EXAMPLE:

a car dealership wants to predict the resale price of a used car.

The dataset contains:

Car Age
Mileage
Engine Size
Number of Owners
Brand
Car Color
Registration Number

```python
import pandas as pd

data = {
    "Car_Age": [2, 4, 5, 7, 9],
    "Mileage": [20000, 40000, 50000, 70000, 90000],
    "Engine_Size": [1200, 1500, 1500, 1800, 2000],
    "Owners": [1, 1, 2, 2, 3],
    "Brand": [1, 2, 1, 3, 2],
    "Color": [1, 2, 3, 1, 2],
    "Registration_No": [101, 102, 103, 104, 105],
    "Price": [800000, 650000, 550000, 400000, 300000]
}

df = pd.DataFrame(data)
# Select useful features
selected = [
    "Car_Age",
    "Mileage",
    "Engine_Size",
    "Owners",
    "Brand"
]
X = df[selected]
print(X)
```
