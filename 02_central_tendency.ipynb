Mean tells us the typical or average value of the data.

For example, if 5 delivery orders took:

20, 25, 30, 35, 40 minutes

the mean tells us the average delivery time.

3. NUMERICAL EXAMPLE:
Suppose a food delivery company records the delivery times of 5 orders:

20, 25, 30, 35, 40 minutes

Step 1: Add all values

20+25+30+35+40 =150

Step 2: Count the values
There are 5 orders.

Step 3: Calculate mean
            150
     mean = --- = 30
             5
So:

Average delivery time = 30 minutes

Python Implementation
Using PYTHON:

```python
delivery_time = [20, 25, 30, 35, 40]

mean = sum(delivery_time) / len(delivery_time)

print("Mean delivery time:", mean, "minutes")
```

```python
import numpy as np

delivery_time = np.array([20, 25, 30, 35, 40])

mean = np.mean(delivery_time)

print("Mean delivery time:", mean)
```

PYTHON IMPLEMENTATION:
Using NUMPY

```python
import numpy as np

delivery_time = np.array([20, 25, 30, 35, 40])

median = np.median(delivery_time)

print("Median delivery time:", median, "minutes")
```

MODE:

1. DEFINITION:
   
Mode is the value that appears most frequently in a dataset.

In simple words:

Mode = The value that occurs the most number of times.

Unlike the mean and median, mode is especially useful for finding the most common choice or category.

2. FORMULA:
   
There is no mathematical calculation like mean.

We simply find the value with the highest frequency.

Mode = Value with the highest frequency

For example:

10, 20, 20, 30, 40

Here, 20 appears twice, while the other values appear once.

Therefore:

Mode = 20

```python
from collections import Counter

orders = [2, 3, 2, 4, 5, 2, 3]

frequency = Counter(orders)

mode = frequency.most_common(1)[0][0]

print("Mode:", mode)
```

Using PANDAS:

```python
import pandas as pd

orders = pd.Series([2, 3, 2, 4, 5, 2, 3])

print("Mode:", orders.mode().iloc[0])
```

```python
import numpy as np

marks = np.array([70, 80, 90])
weights = np.array([0.20, 0.30, 0.50])

weighted_mean = np.average(marks, weights=weights)

print("Weighted Mean:", weighted_mean)
```

USING NUMPY:

```python
import numpy as np

growth_factors = [1.20, 1.50]

geometric_mean = np.prod(growth_factors) ** (1 / len(growth_factors))

growth_rate = (geometric_mean - 1) * 100

print("Geometric Mean:", geometric_mean)
print("Average Growth Rate:", growth_rate, "%")
```

HARMONIC MEAN:

FORMULA:

For two values 
x and y:

        2xy
   HM = ---
        x+y

For n values:

                n

```python
from scipy.stats import hmean

speeds = [30, 60]

result = hmean(speeds)

print("Harmonic Mean:", result)
```
