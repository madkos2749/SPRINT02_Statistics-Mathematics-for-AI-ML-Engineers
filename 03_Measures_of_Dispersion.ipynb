RANGE:

DEFINITION:
Range is the simplest measure of dispersion. It tells us how far apart the smallest and largest values in a dataset are.

In simple real words:

Range = Biggest value − Smallest value

It helps us understand whether the data is closely grouped or widely spread.

FORMULA:

RANGE = Maximum Value - Minimum Value

Example:
Maximum Value = 80
Minimum Vlaue = 40
RANGE = 80 - 40
RANGE = 40

EXAMPLE:
USING NUMPY:

```python
import numpy as np

delivery_time = np.array([20, 25, 30, 35, 45])

range_value = np.max(delivery_time) - np.min(delivery_time)

print("Range:", range_value, "minutes")
```

DEFINITION:
Variance is a measure of dispersion that tells us how far the data values are spread out from their average (mean).

In simple real words:

Variance tells us how much the values differ from the average value.

Small variance → values are close to the mean.

Large variance → values are far away from the mean.

Real-time Example:

Suppose a food delivery company records the delivery times of 4 orders:

20, 22, 24, 26

Mean = 20 + 22 + 24 + 26 / 4 
Mean = 23

Step-2:
Find the difference from Mean:

Delivery Time  | 	Difference from Mean  |	Squared Difference
------------------------------------------------------------------
20	           |          -3	          |          9
22             |        -1	              |          1
24	           |          +1              |         1
26	           |          +3              |        	9

STEP-3:
Calculate Variance for above example;
  Variance = 9 + 1 + 1 + 9/4
  Variance = 20/ 4
  Variance  = 5

PYTHON EXAMPLE:
USING NUMPY:

```python
import numpy as np

delivery_time = np.array([20, 22, 24, 26])

variance = np.var(delivery_time)

print("Variance:", variance)
```

Variance = How much the data spreads around the mean.

Think of throwing balls at a target:

Balls close together → Low variance

Balls spread far apart → High variance

One-line definition
Variance is a measure of dispersion that calculates how far data values are spread around their mean by averaging their squared differences from the mean.

STANDARD DEVIATION:

DEFINITION:
Standard Deviation is a measure of dispersion that tells us how much the data values typically spread away from the mean (average).

Standard deviation tells us whether the data values are usually close to the average or far away from it.

Small standard deviation → values are close to the average.

Large standard deviation → values are more spread out.

Note: Standard deviation is a measure of dispersion, not a measure of central tendency.

EXAMPLE:
Suppose a food delivery company wants to record delivery times;

20, 22, 24, 26

mean = 20 + 22 + 24 + 26 / 4 =23
so, mean = 23

The population variance is 5:

Standard Deviation = square root of 5 ~~ 2.24

PYTHON EXAMPLE:

```python
import numpy as np

delivery_time = [20, 22, 24, 26]

std = np.std(delivery_time)

print("Standard Deviation:", std)
```

NOTE: Standard deviation is the square root of variance and tells us how much data values typically differ from their mean.

QUARTILES:

quartiles themselves are measures of position, but they are used to understand dispersion (spread), especially through the Interquartile Range (IQR).

Quartiles divide ordered data into four parts and help us understand how the data is spread.

The important quartiles are:

Q1 → 25th percentile

Q2 → 50th percentile (Median)

Q3 → 75th percentile

The main measure of dispersion based on quartiles is:

IQR = Q3 − Q1

IQR tells us how spread out the middle 50% of the data is.

Real-time EXAMPLE:


10, 15, 20, 25, 30, 35, 40, 45

Using Python we get approximately:

Q1 = 17.5
Q2 = 27.5
Q3 = 37.5

Now calculate the IQR:

IQR = Q3 − Q1

IQR = 37.5 − 17.5

IQR = 20

This means; The middle 50% of delivery times are spread across 20 minutes.

So, instead of looking at the minimum and maximum values, IQR focuses on the middle part of the data.

NUMPY EXAMPLE:

```python
import numpy as np

delivery_time = [10, 15, 20, 25, 30, 35, 40, 45]

Q1 = np.percentile(delivery_time, 25)
Q3 = np.percentile(delivery_time, 75)

IQR = Q3 - Q1

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
```

DEFINITION:

Percentile tells us the position of a particular value compared with the rest of the data.

In simple real words:

A percentile tells us what percentage of data values are below or at a particular value.

For example, if a student's score is at the 80th percentile, it means the student scored better than or equal to about 80% of the students.

Percentiles are mainly measures of position, but they are useful for understanding the spread and distribution of data.

EXAMPLE:
Suppose a food delivery company records delivery times:

The company wants to know:

"What is the 90th percentile delivery time?"

Using the 90th percentile, we get approximately:

90th percentile = 51 minutes.

PYTHON EXAMPLE:
USING NUMPY:

```python
import numpy as np

delivery_time = [10, 15, 20, 25, 30, 35, 40, 45, 50, 60]

p90 = np.percentile(delivery_time, 90)

print("90th Percentile:", p90, "minutes")
```

NOTE: A percentile indicates the percentage of data values that are less than or equal to a particular value and helps us understand the position and spread of data in AI/ML.

INTERQUARTILE RANGE [IQR]: 

DEFINTION:
Interquartile Range (IQR) is a measure of dispersion that tells us how spread out the middle 50% of the data is.

In simple real words:

IQR shows the distance between the 25th percentile (Q1) and the 75th percentile (Q3).

It is useful because it focuses on the middle part of the data and is less affected by extreme values or outliers.

FORMULA:

        IQR = Q3 - Q1
Where, 
Q1 = 25th percentile

Q3 = 75th percentile

What does it mean:

The middle 50% of delivery times are spread across 20 minutes.

So IQR tells the company about the typical spread of the central part of its delivery data, without being strongly influenced by very fast or very slow deliveries.

PYTHON EXAMPLE:
USING NUMPY:

```python
import numpy as np

delivery_time = [10, 15, 20, 25, 30, 35, 40, 45]

Q1 = np.percentile(delivery_time, 25)
Q3 = np.percentile(delivery_time, 75)

IQR = Q3 - Q1

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
```

NOTE: Interquartile Range (IQR) is the difference between Q3 and Q1 and measures how spread out the middle 50% of a dataset is.

COEFFICIENT OF VARIATION:

DEFINITION:

Coefficient of Variation (CV) is a measure of dispersion that tells us how much variation there is compared with the average (mean).

In simple real words:

CV tells us how large the spread of the data is compared with its average value.

It is especially useful when we want to compare the variability of two datasets that have different averages or different units.

FORMULA:

                 STANDARD DEVIATION
           CV = -------------------- * 100
                       MEAN

The answer is usually expressed as a percentage.

Lower CV → data is more consistent.

Higher CV → data is more variable.

EXAMPLE:
Suppose a delivery company records delivery times:

20, 22, 24, 26, 28 minutes

Step 1: Find the mean

Mean = 20+22+24+26+28/5

Mean = 24

Step 2: Find standard deviation
The population standard deviation is approximately:

SD ≈ 2.83

Step 3: Calculate CV
CV = 2.83 / 24×100

CV ≈ 11.8%

Interpretation
A CV of approximately 11.8% means the delivery times have relatively low variation compared with their average.

So, the delivery service is fairly consistent.

PYTHON EXAMPLE:
USING NUMPY:

```python
import numpy as np

delivery_time = np.array([20, 22, 24, 26, 28])

mean = np.mean(delivery_time)
std = np.std(delivery_time)

cv = (std / mean) * 100

print("Mean:", mean)
print("Standard Deviation:", std)
print("Coefficient of Variation:", cv, "%")
```

NOTE: Coefficient of Variation (CV) is a relative measure of dispersion that expresses the standard deviation as a percentage of the mean, helping us compare variability between datasets.
