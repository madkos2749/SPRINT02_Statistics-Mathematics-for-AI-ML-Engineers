Inferential Statistics:

Inferential Statistics means using a small amount of data (sample) to make a conclusion or prediction about a larger group (population).

In very simple words:

Inferential Statistics = Study a small group and use it to understand a bigger group.

EXAMPLE:

Imagine a company has 100,000 customers.

The company wants to know:

"Are our customers satisfied with our product?"

It is difficult to ask all 100,000 customers.

So, the company randomly selects 500 customers and asks them.

Suppose:

400 out of 500 customers say "Yes, I am satisfied."

100 say "No."

So:

400
--- * 100 = 80%
500

The company can estimate that around 80% of all customers may be satisfied.

Here:

100,000 customers → Population

500 customers → Sample

80% satisfied → Result from the sample

POPULATOIN:

Population means the complete group of people, objects, or data that we are interested in studying.

EXAMPLE:

There may be 1 million houses in your target market.

If your goal is to predict prices for all those houses:

1 million houses = Population

A sample is a smaller part of a large group of data that we select and study.

The large group is called the population, and the smaller part we take from it is called the sample.

EXAMPLE:

Suppose a food delivery company has 1,00,000 customers.

The company wants to know:

"How happy are our customers with our delivery time?"

It is difficult to ask all 1,00,000 customers.

So, the company randomly selects 1,000 customers and asks them about their experience.

Here:

1,00,000 customers → Population

1,000 selected customers → Sample

Their answers → Sample data

For example, if 800 out of 1,000 customers say they are happy:

800
---- * 100 = 80%
1000

The company can use this sample result to estimate the satisfaction of the larger customer population.

How this relates to AI/ML:
Suppose the company wants to build an ML model that predicts whether a customer will be satisfied.

It may have millions of customer records, but it could use a selected portion of the data for analysis or training.

Uses of Sample:

Using a sample can make the work:

1. Faster

2. Cheaper

3. Easier to analyze

4. Practical when the population is very large

EXAMPLE:

"What is the average delivery time?"

Checking all 10,000 orders may take more time. So, it randomly selects 100 orders and checks their delivery times.

```python
import pandas as pd

# All delivery orders
orders = pd.DataFrame({
    "order_id": range(1, 11),
    "delivery_time": [25, 30, 28, 35, 40, 32, 27, 45, 31, 29]
})

# Select a random sample of 3 orders
sample = orders.sample(n=3, random_state=42)

print("Selected Sample:")
print(sample)

# Calculate average delivery time of the sample
average_time = sample["delivery_time"].mean()

print("Average delivery time:", average_time)
```

A feature is a useful piece of information that an ML model uses to learn and make a prediction.

```python
import pandas as pd

data = pd.DataFrame({
    "distance_km": [2, 5, 3, 8],
    "rain": [0, 1, 0, 1],
    "restaurant_busy": [0, 1, 1, 1],
    "delivery_time": [20, 45, 30, 60]
})

# Features
X = data[["distance_km", "rain", "restaurant_busy"]]

# Target
y = data["delivery_time"]

print("Features:")
print(X)

print("\nTarget:")
print(y)
```
