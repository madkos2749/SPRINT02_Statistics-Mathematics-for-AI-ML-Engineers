# CORRELATION & COVARIANCE

## CORRELATION:

### DEFINITION:

Correlation is a statistical measure that tells us the strength and direction of the relationship between two variables.

The correlation coefficient usually ranges from -1 to +1:

+1 → Perfect positive correlation

0 → No linear correlation

-1 → Perfect negative correlation

In simple words: Correlation tells us whether two variables tend to increase or decrease together.

##### Real time Examples:

* Healthcare: Analyze the relationship between exercise and heart rate.

* Business: Check the relationship between advertising spending and sales.

* Finance: Study relationships between stock prices.

* Weather: Analyze the relationship between temperature and electricity usage.

#### EXAMPLE:
Suppose an ice cream shop wants to know whether temperature affects ice cream sales.

As the temperature increases, people generally buy more ice cream.

```python
import pandas as pd

data = {
    "Temperature": [20, 24, 28, 32, 36],
    "IceCream_Sales": [40, 55, 75, 100, 130]
}

df = pd.DataFrame(data)

correlation = df["Temperature"].corr(df["IceCream_Sales"])

print("Correlation:", round(correlation, 2))
```

##### Code Explanation:

The result shows how the one cause relates to increase the another cause to happen, which is almost equal to ~1.
Correlation shows that two variables are related; it does not by itself prove that one causes the other.

## COVARIANCE:

### DEFINITION:

Covariance is a statistical measure that tells us how two variables change together.

In simple words: Covariance tells us whether two variables tend to increase together or move in opposite directions.

* Positive covariance → Both variables tend to increase or decrease together.
* Negative covariance → When one increases, the other tends to decrease.
* Zero covariance → No clear linear relationship.

##### Real time Examples:

* Finance: Analyze how two stock returns move together.
* Healthcare: Study how two health measurements vary together.
* Business: Analyze the relationship between advertising expenditure and sales.
* Weather: Study how temperature and electricity consumption change together.

#### EXAMPLE:

Suppose a shop records rainfall and umbrella sales for 5 days.

```python
import numpy as np

rainfall = [2, 5, 8, 10, 15]
sales = [10, 20, 30, 40, 50]

covariance = np.cov(rainfall, sales)[0, 1]

print("Covariance:", covariance)
```

##### Code Explanation:

As rainfall increases, umbrella sales also increase.
Therefore, the covariance is positive.

Because the covariance is positive, rainfall and umbrella sales generally move in the same direction.


## POSITIVE CORRELATION:

### DEFINITION:

Positive correlation is a relationship between two variables where both variables tend to increase together or decrease together.
It occurs when two variables tend to move in the same direction. When one variable increases, the other also tends to increase; when one decreases, the other tends to decrease.

The correlation coefficient for positive correlation is between 0 and +1.

* +1 → Perfect positive correlation

* Close to +1 → Strong positive correlation

* Close to 0 → Weak positive correlation

##### Real time Examples:

* Education: More study hours → Higher marks
* Business: More advertising → Higher sales
* Weather: Higher temperature → More cold-drink sales
* Fitness: More exercise → More calories burned

#### EXAMPLE:

Suppose a fitness app records the following:

Exercise Time (minutes) and Calories Burned

```python
import pandas as pd

data = {
    "Exercise_Minutes": [10, 20, 30, 40, 50],
    "Calories": [60, 120, 180, 240, 300]
}

df = pd.DataFrame(data)

result = df["Exercise_Minutes"].corr(df["Calories"])

print("Positive Correlation:", result)
```

##### Code Explanation:
The result is +1, meaning these example variables have a perfect positive correlation.

## NEGATIVE CORRELATION:

### DEFINITION:

Negative correlation occurs when two variables tend to move in opposite directions. When one variable increases, the other generally decreases, and vice versa.

The correlation coefficient for negative correlation is between -1 and 0.

* -1 → Perfect negative correlation
* Close to -1 → Strong negative correlation
* Close to 0 → Weak negative correlation

##### Real time Examples:

* Weather: Temperature ↑ → Heater usage ↓
* Fitness: Exercise ↑ → Body fat may ↓
* Business: Product price ↑ → Demand may ↓
* Traffic: Traffic speed ↑ → Travel time ↓

#### EXAMPLE:

Suppose a smart-home system records:

Temperature (°C) and Heater Usage (minutes)

```python
import pandas as pd

data = {
    "Temperature": [10, 15, 20, 25, 30],
    "Heater_Minutes": [60, 50, 40, 25, 10]
}

df = pd.DataFrame(data)

result = df["Temperature"].corr(df["Heater_Minutes"])

print("Negative Correlation:", round(result, 2))
```

##### Code Explanation:
The result is close to -1, so there is a very strong negative correlation between temperature and heater usage in this example.

## NO CORRELATION:

### DEFINITION:

No correlation means there is no meaningful linear relationship between two variables. A change in one variable does not consistently correspond to a change in the other.

The two variables do not have a predictable linear relationship.

The correlation coefficient is close to 0.

* +1 → Perfect positive correlation

* 0 → No linear correlation

* -1 → Perfect negative correlation

##### Real time Examples:

* Education: Shoe size and exam marks may have little relationship.
* Business: Employee shoe size and monthly sales may have no meaningful relationship.

## PEARSON CORRELATION:

### DEFINITION:

Pearson correlation is a statistical method used to measure the strength and direction of the linear relationship between two numerical variables.

It is represented by r and its value ranges from -1 to +1.

* +1 → Perfect positive linear relationship
* 0 → No linear relationship
* -1 → Perfect negative linear relationship


In simple words: Pearson correlation tells us how strongly two numerical variables are linearly related.


##### Real time Examples:

* Healthcare: Study relationships between age and blood pressure.
* Finance: Analyze relationships between investment returns.
* AI/ML: Identify relationships between numerical features and help with feature selection.

#### EXAMPLE:

Screen Time vs Phone Battery Usage
Suppose we want to find the relationship between screen time and battery usage.

###### Screen Time (hours) |	 Battery Used (%)
---------------------------------------------------
1	                                   12
2	                                   25
3	                                   38
4	                                   52
5	                                   65

```python
import pandas as pd

data = {
    "Screen_Time": [1, 2, 3, 4, 5],
    "Battery_Used": [12, 25, 38, 52, 65]
}

df = pd.DataFrame(data)

r = df["Screen_Time"].corr(df["Battery_Used"])

print("Pearson Correlation:", round(r, 2))
```

##### Code Explanation:

The value is very close to +1, indicating a strong positive linear relationship in this illustrative data.

## SPEARMAN CORRELATION:

### DEFINITION:

Spearman correlation is a statistical method used to measure the strength and direction of the relationship between two variables based on their ranks.

It is represented by ρ (rho) and ranges from -1 to +1.

+1 → Perfect positive relationship

0 → No monotonic relationship

-1 → Perfect negative relationship

##### Real time Examples:

* Education: Compare students' study-time rankings with their exam-score rankings.
* Customer reviews: Compare product rankings with customer satisfaction rankings.
* Sports: Compare players' rankings in different performance categories.

#### EXAMPLE:

Suppose we ask five people to rank their preference for cold drinks on different days based on temperature.

```python
from scipy.stats import spearmanr

temperature = [20, 24, 28, 32, 36]
preference = [1, 2, 3, 4, 5]

rho, _ = spearmanr(temperature, preference)

print("Spearman Correlation:", rho)
```

##### Code Explanation:
+0.999999999999999 is closely to +1.0, which indicates a perfect positive rank relationship in this example.

## CORRELATION MATRX

### DEFINITION:

A correlation matrix is a table that shows the correlation coefficients between multiple variables at the same time.

Each value usually ranges from -1 to +1:

* +1 → Perfect positive correlation
* 0 → No linear correlation
* -1 → Perfect negative correlation

The diagonal values are always 1, because every variable has a perfect correlation with itself.

##### Real time Examples:

AI/ML: Understanding relationships between features.

* Feature selection: Finding highly related or redundant features.
* PCA: Studying relationships among variables before dimensionality reduction.

#### EXAMPLE:

**FITNESS DATA**

Suppose a fitness app records:

Exercise minutes

Calories burned

Water intake

We can use a correlation matrix to see how these variables are related.

```python
import pandas as pd

data = {
    "Exercise": [10, 20, 30, 40, 50],
    "Calories": [50, 100, 150, 200, 250],
    "Water": [2, 2.5, 3, 3.5, 4]
}

df = pd.DataFrame(data)

print(df.corr().round(2))
```

##### Code Explanation:

So, all three variables have a perfect positive correlation in this small example.

Easy to remember: A correlation matrix lets us see the relationships between many variables at once, instead of calculating each correlation separately.


## *HEATMAP:*

### DEFINITION:

A heatmap is a graphical representation of data where colors are used to show the magnitude or intensity of values.

It uses different colors to make patterns and high/low values easy to understand.

##### Real time Examples:

* Websites: Show where users click or spend time on a webpage.
* Sports: Show player activity or shot locations on a field.
* Geography: Display population density or traffic intensity.

#### EXAMPLE:

import seaborn as sns
import matplotlib.pyplot as plt

sales = [
    [30, 35, 40, 38, 55],
    [45, 50, 48, 60, 90]
]

sns.heatmap(
    sales,
    annot=True,
    cmap="YlOrRd",
    xticklabels=["Mon", "Tue", "Wed", "Thu", "Fri"],
    yticklabels=["Lunch", "Evening"]
)

plt.title("Restaurant Sales Heatmap")
plt.show()


```python
import seaborn as sns
import matplotlib.pyplot as plt

study = [
    [2, 3, 1, 4, 5],   # Maths
    [1, 2, 3, 2, 4],   # Science
    [4, 3, 2, 5, 3]    # English
]

sns.heatmap(
    study,
    annot=True,
    cmap="YlGnBu",
    xticklabels=["Mon", "Tue", "Wed", "Thu", "Fri"],
    yticklabels=["Maths", "Science", "English"]
)

plt.title("Weekly Study Heatmap")
plt.xlabel("Day")
plt.ylabel("Subject")
plt.show()
```

##### Code Explanation:

1 hour  → Light color  🟦
2 hours → Slightly darker
3 hours → Medium
4 hours → Dark
5 hours → Very dark     🟩


So, by looking at the heatmap, we can immediately see that:

* Maths on Friday = 5 hours → highest study time
* Maths on Wednesday = 1 hour → lowest udy time
* English on Thursday = 5 hours → highest study time


A heatmap represents numerical data using colors, where different color intensities represent different values.
