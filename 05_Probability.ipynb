EVENTS

DEFINITION:

An event in probability is a specific outcome or a group of possible outcomes that we are interested in.
An event is something that we want to check whether it happens or not.

EXAMPLE:

when rolling a dice:

Getting 6 → an event

Getting an even number (2, 4, 6) → an event

Getting a number greater than 4 (5, 6) → an event

Possible outcomes:

Sample Space = {1, 2, 3, 4, 5, 6}

Let event A = "getting an even number."

Event A = {2, 4, 6}

Therefore:

P(A)=3/6=0.5

So, the probability of getting an even number is 50%.

```python
import random

orders = 10
delayed = 0

for order in range(1, orders + 1):
    is_delayed = random.choice([True, False])

    if is_delayed:
        delayed += 1
        print(f"Order {order}: Delayed")
    else:
        print(f"Order {order}: On time")

print("Estimated probability of delay:", delayed / orders)
```

Here, 3 out of 10 orders were delayed.

So the observed probability is:

P(Delay) = 3/10 = 0.3 = 30%

SAMPLE SPACE:

DEFINITION:

Sample Space is the complete set of all possible outcomes that can happen in a probability experiment.

In simple words:

Sample Space = all possible things that could happen.

It is usually represented by S or Ω (omega).

EXAMPLE:

Suppose we toss a coin once.

There are only two possible outcomes:

S = {Heads, Tails}

So, the sample space contains all possible results of the experiment.

Another example: rolling a dice:

S = {1, 2, 3, 4, 5, 6}

Here, the sample space has 6 possible outcomes.

Imagine a food-delivery app wants to predict the status of an order.

An order can have these possible statuses:

S = {
    "Preparing",
    "Out for Delivery",
    "Delivered",
    "Cancelled"
}

```python
import random

sample_space = [
    "Preparing",
    "Out for Delivery",
    "Delivered",
    "Cancelled"
]

order_status = random.choice(sample_space)

print("Possible statuses:", sample_space)
print("Current order status:", order_status)
```

NOTE:
Sample Space = all possible outcomes. Event = a particular outcome or group of outcomes from that sample space.

INDEPENDENT EVENTS:

DEFINITION:

Independent events are events where the occurrence of one event does not affect the probability of the other event.

In simple words:
If one event happens, it does not change the chance of the other event happening.

FORMULA:

P(A∩B) = P(A) × P(B)

EXAMPLE:

Suppose we:

Toss a coin.

Roll a dice.

Getting Heads on the coin does not affect the number obtained on the dice.

Let:

Event A = Getting Heads → P(A) = 1/2

Event B = Getting 6 → P(B) = 1/6

Therefore:

P(A∩B) = 1/2 × 1/6 = 1/12

So, the chance of getting Heads AND 6 is 1/12.

EXAMPLE:

Imagine a smart factory has two separate sensors:

Sensor A: Detects high temperature.

Sensor B: Detects high vibration.

```python
import random

temperature_alert = random.random() < 0.3
vibration_alert = random.random() < 0.2

print("Temperature alert:", temperature_alert)
print("Vibration alert:", vibration_alert)
```

random.random() < 0.3

means there is approximately a 30% chance of a temperature alert.

random.random() < 0.2

means there is approximately a 20% chance of a vibration alert.

If we want both alerts to happen:

P(A∩B) = 0.3 × 0.2 = 0.06

So the probability is:

0.06 = 6%

Temperature alert → 30%
Vibration alert   → 20%

Both alerts       → 30% × 20%
                  → 6%

NOTE:
Independent events = one event happening does not change the chance of the other event.

DEPEDENT EVENT:

DEFINITION:
Dependent events are events where the occurrence of one event changes the probability of another event.

In simple words:

If one event happens, it affects the chance of the next event happening.

FORMULA:

P(A∩B) = P(A) × P(B∣A)

EXAMPLE:
Suppose a box contains:

3 Red balls
2 Blue balls

We pick two balls without replacing the first ball.

Event A → First ball is Red

Event B → Second ball is Red

Initially:

P(A) = 3|5

If the first ball was Red, only 2 Red balls remain out of 4:

P(B ∣ A) = 2|4

Therefore:

P(A∩B) = 3|5 × 2|4 = 3|10

The first selection changed the probability of the second selection, so these are dependent events.



EXAMPLE:

Imagine an e-commerce website.

A customer:

Views a product

Then may add that product to the cart

These events are dependent because viewing the product can affect the chance of adding it to the cart.

```python
import random

viewed = random.random() < 0.8

if viewed:
    added_to_cart = random.random() < 0.4
else:
    added_to_cart = random.random() < 0.1

print("Product viewed:", viewed)
print("Added to cart:", added_to_cart)
```

CODE EXPLANATION:

First:
viewed = random.random() < 0.8

means the customer has an 80% chance of viewing the product.

Then the probability of adding to the cart depends on whether they viewed it:

if viewed:
    added_to_cart = random.random() < 0.4
else:
    added_to_cart = random.random() < 0.1

So:

Product viewed       → Add to cart: 40%
Product not viewed   → Add to cart: 10%

NOTE:
the first event changes the probability of the second event, they are dependent events.

CONDITIONAL PROBABILITY:

DEFINITION:

Conditional probability is the probability of an event happening when we already know that another event has happened.

In simple words:

Conditional probability means: “What is the chance of A, given that we already know B?”

It is written as:

P(A ∣ B)

P(A | B) = P(A∩B)/P(B)

EXAMPLE:

Imagine an email system trying to identify spam emails.

The system notices that an email contains the word "FREE".

Now it asks:

What is the probability that this email is spam, given that it contains "FREE"?

This is conditional probability:

P(Spam ∣ FREE)

```python
import random

emails = 100
free_emails = 20
spam_free_emails = 15

probability = spam_free_emails / free_emails

print("P(Spam | FREE) =", probability)
```

CODE EXPLANATION:
-----------------

Suppose we have:

100 total emails
       ↓
20 contain "FREE"
       ↓
15 of those are spam

We are not looking at all 100 emails anymore.

We already know the email contains "FREE", so we only look at those 20 emails.

Probability = Spam emails containing FREE
              ──────────────────────────────
              Emails containing FREE

            = 15 / 20

            = 75%

P(Spam | "FREE") → Spam detection

P(Buy | Viewed product) → Customer prediction

P(Disease | Symptoms) → Medical prediction

P(Fraud | Unusual transaction) → Fraud detection

NOTE: 
Conditional probability = the probability of something happening when we already know some related information.

BAYE'S THEOREM:

DEFINITION:

Bayes' Theorem is a method for updating the probability of something when we get new information.

In simple words:

Bayes' Theorem tells us how our belief about something should change after seeing new evidence.

FORMULA:

                                  P(A ∣ B) = P(B ∣ A) × P(A)
                                  -------------------------
                                            P(B)

Where:

P(A) → probability of A before seeing new information

P(B|A) → probability of seeing B if A is true

P(B) → probability of seeing B

P(A|B) → updated probability of A after seeing B

EXAMPLE:

Imagine a box contains:

10 balls
3 Red
7 Blue

So initially:

P(Red) = 3/10 = 30%

Now suppose we receive some additional information that changes what we know about the box. Bayes' theorem lets us update our probability using that new evidence.

EXAMPLE:

Imagine a bank's ML system wants to identify whether a transaction is fraudulent.

Suppose:

Only 1% of transactions are normally fraud.

If a transaction is actually fraud, the fraud detector flags it 90% of the time.

The detector also incorrectly flags 5% of genuine transactions.

We want to know:

If a transaction is flagged, what is the probability that it is actually fraud?

Using Bayes' theorem:

P(Fraud∣Flag)=P(Flag∣Fraud)P(Fraud)P(Flag)

First calculate:

P(Flag)=(0.90×0.01)+(0.05×0.99)

P(Flag)=0.0585

Therefore:

P(Fraud ∣ Flag)= 0.90 × 0.01/0.0585≈0.154

So the probability is approximately 15.4%.

This is an important AI/ML idea: even a good detector can produce many false alarms when the actual event is rare.

```python
fraud = 0.01
flag_if_fraud = 0.90
flag_if_genuine = 0.05
genuine = 0.99

flag = (flag_if_fraud * fraud) + (flag_if_genuine * genuine)

prob_fraud_if_flagged = (flag_if_fraud * fraud) / flag

print("P(Fraud | Flag) =", prob_fraud_if_flagged)
```

CODE EXPLANATION:

The code first calculates how likely a transaction is to be flagged:

flag = (flag_if_fraud * fraud) + (flag_if_genuine * genuine)

Then Bayes' theorem updates the probability:

prob_fraud_if_flagged = (flag_if_fraud * fraud) / flag

The key point is that we started with only a 1% fraud rate, then received new evidence: the transaction was flagged.

Bayes' theorem tells us how to update our belief to about 15.4%.

Uses of Bayes' theorem :

Spam detection

Fraud detection

Medical diagnosis

Recommendation systems

Classification

Naive Bayes algorithms


NOTE:
Bayes' Theorem = Start with what you know → receive new evidence → update the probability.

PROBABILITY RULES:

DEFINITION:

Probability rules are basic mathematical rules used to calculate the chance of events happening.

In simple words:

Probability rules help us combine, compare, and update the chances of different events.

These rules are important in AI/ML because ML models often work with uncertain outcomes.

1. Addition Rule — OR
   ------------------
When we want the probability of A or B happening:

P(A∪B)=P(A)+P(B)−P(A∩B)

For events that cannot happen together, simply:

P(A∪B)=P(A)+P(B)

Example: A dice is rolled. What is the chance of getting 1 or 6?

P(1 or 6)=
1/6 + 1/6 = 2/6 = 1/3


So the probability is 33.3%.

2. Multiplication Rule — AND
   -------------------------
For independent events:

P(A∩B)=P(A)×P(B)

Example: Toss a coin and roll a dice.

Probability of getting Heads AND 6:

1/2×1/6=1/12

So the probability is about 8.33%.

3. Complement Rule — NOT
   ----------------------
   
If the probability of an event happening is known, the probability of it not happening is:

P(Not A)=1−P(A)

Example:

If the probability that an online order is late is 0.2:

P(Not Late)=1−0.2=0.8

So, there is an 80% chance that it is not late.

4. Conditional Probability — GIVEN
   -------------------------------
   
Sometimes we already know that something happened.

P(A∣B)=P(A∩B)/P(B)

It means:

What is the probability of A, given that B has already happened?

For example:

P(Spam | Email contains "FREE")

This is commonly useful in spam classification.

NOTE:
Probability rules help AI/ML systems calculate the likelihood of events and make better decisions when outcomes are uncertain.

RANDOM VARIABLES:

DEFINITION:

A random variable is a variable that represents the outcome of a random event using a value.

In simple words:

A random variable is a number or value assigned to the result of something uncertain.

It is usually represented by a capital letter such as X.

Example:

Suppose we toss a coin.

The outcomes are:

Heads
Tails

We can define a random variable X as:

X = 1 → Heads
X = 0 → Tails

Now, whenever we toss the coin, X gets either 0 or 1.

For example:

Toss 1 → Heads → X = 1
Toss 2 → Tails → X = 0
Toss 3 → Heads → X = 1

EXAMPLE:

```python
import random

for order in range(1, 6):
    X = random.choice([0, 1])

    if X == 1:
        print("Order", order, "→ Delayed (X = 1)")
    else:
        print("Order", order, "→ On time (X = 0)")
```

Two Types of Random Variables:
1. Discrete Random Variables
2. Continuous Random Variables

1. DISCRETE RANDOM VARIABLES:

Has countable values.

Examples:

Number of customers
Number of defective products
Number of clicks
Number of orders

For example:

X = Number of orders today
X = {0, 1, 2, 3, ...}

2. CONTINUOUS RANDOM VARIABLES:

Can take any value within a range.

Examples:

Delivery time
Temperature
Height
Weight

For example:

X = Delivery time
X = 32.5 minutes

 NOTE:
 A random variable converts the uncertain outcome of a real-world event into a value that we can analyze mathematically and use in AI/ML.

EXPECTED VALUE:

Expected Value (EV) is the average result we expect over many repeated trials, based on the probability of each possible outcome.

In simple words:

Expected value tells us what outcome we can expect on average in the long run.

DEFINITION:

Expected Value (EV) is the average result we expect over many repeated trials, based on the probability of each possible outcome.

In simple words:

Expected value tells us what outcome we can expect on average in the long run.

DEFINITION:

E(X)=∑xP(X=x)


EXAMPLE:
Imagine a delivery company wants to estimate its average delay cost.

Suppose an order has:

Outcome	     |   Cost  |	Probability
------------------------------------
On time	     |    ₹0   |      0.7
Small delay	 |   ₹100  |      0.2
Large delay	 |   ₹500  |	  0.1

                                      cost: E(X)=(0×0.7)+(100×0.2)+(500×0.1)
                                    -->  E(X)=0+20+50=₹70

```python
costs = [0, 100, 500]
probabilities = [0.7, 0.2, 0.1]

expected_cost = sum(
    cost * probability
    for cost, probability in zip(costs, probabilities)
)

print("Expected cost per order: ₹", expected_cost)
```

NOTE:
Expected Value = the probability-weighted average of all possible outcomes.
